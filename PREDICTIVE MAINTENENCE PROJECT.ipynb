## 5.2.6 Feature Scaling

Machine learning algorithms work with numerical values. However, different features in a dataset often have different units and ranges.

For example:

- Rotational Speed → 1200–2800 RPM
- Torque → 3–80 Nm
- Tool Wear → 0–250 Minutes
- Air Temperature → 295–305 K

If these features are used directly, variables with larger numerical values may dominate the learning process.

To avoid this issue, numerical features are standardized using **StandardScaler**.

StandardScaler transforms each feature so that it has:

- Mean = 0
- Standard Deviation = 1

Feature scaling improves the performance of distance-based and optimization-based machine learning algorithms.

## Introduction

K-Nearest Neighbors (KNN) is a supervised machine learning algorithm used for both classification and regression tasks.

In this project, KNN is used as a classification algorithm to predict whether a manufacturing machine is likely to experience a failure based on its operating conditions.

Unlike many machine learning algorithms, KNN does not build an explicit mathematical model during training. Instead, it stores the training data and classifies new observations by identifying the K most similar training samples.

Similarity between observations is measured using a distance metric, most commonly the Euclidean Distance.

Because KNN is a distance-based algorithm, feature scaling is essential to ensure that all variables contribute fairly to the distance calculation.

## Business Objective

The objective of applying the K-Nearest Neighbors algorithm is to classify machines as either:

- No Failure (0)
- Machine Failure (1)

using historical manufacturing sensor data.

The model will analyze machine operating parameters such as:

- Product Type
- Air Temperature
- Process Temperature
- Rotational Speed
- Torque
- Tool Wear

to predict whether a machine is likely to fail.

An accurate prediction model can support predictive maintenance by reducing unexpected downtime, improving equipment reliability, and optimizing maintenance schedules.

## 6.3 Hyperparameter Tuning

The performance of the K-Nearest Neighbors algorithm depends heavily on the choice of the hyperparameter **K**, which represents the number of nearest neighbors considered during classification.

A very small K may make the model sensitive to noise, while a very large K may oversimplify the decision boundary.

Therefore, different values of K are evaluated to identify the model that provides the best predictive performance.

````python
from sklearn.metrics import accuracy_score

train_scores = []
test_scores = []

k_values = range(1, 31)

for k in k_values:

    model = KNeighborsClassifier(n_neighbors=k)

    model.fit(X_train_scaled, y_train)

    train_pred = model.predict(X_train_scaled)
    test_pred = model.predict(X_test_scaled)

    train_scores.append(accuracy_score(y_train, train_pred))
    test_scores.append(accuracy_score(y_test, test_pred))
````

````python
knn_results = pd.DataFrame({
    "K Value": list(k_values),
    "Training Accuracy": train_scores,
    "Testing Accuracy": test_scores
})
accuracy_df = knn_results
accuracy_df
````

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

plt.plot(k_values, train_scores,
         marker='o',
         label="Training Accuracy")

plt.plot(k_values, test_scores,
         marker='s',
         label="Testing Accuracy")

plt.title("KNN Hyperparameter Tuning")

plt.xlabel("Number of Neighbors (K)")

plt.ylabel("Accuracy")

plt.xticks(range(1,31))

plt.grid(True)

plt.legend()

plt.show()
````

````python
best_index = knn_results["Testing Accuracy"].idxmax()

best_k = knn_results.loc[best_index, "K Value"]

best_accuracy = knn_results.loc[best_index, "Testing Accuracy"]

print(f"Best K Value : {best_k}")

print(f"Best Testing Accuracy : {best_accuracy:.4f}")
````

Hyperparameter tuning is the process of finding the optimal combination of model parameters that improves predictive performance.

````python
import time
from sklearn.linear_model import LogisticRegression
solvers = ['liblinear', 'lbfgs', 'newton-cg', 'sag', 'saga']

lr_results = []

for solver in solvers:

    start = time.time()

    model = LogisticRegression(
        solver=solver,
        max_iter=1000,
        random_state=42
    )

    model.fit(X_train_scaled, y_train)

    end = time.time()

    y_pred = model.predict(X_test_scaled)

    lr_results.append({
        "Solver": solver,
        "Testing Accuracy": round(accuracy_score(y_test, y_pred), 4),
        "Training Time (sec)": round(end - start, 4),
        "Iterations": model.n_iter_[0]
    })

solver_results = pd.DataFrame(lr_results)

print(solver_results)
````

A Decision Tree is a supervised machine learning algorithm used for both classification and regression problems.

It predicts the target variable by learning a sequence of decision rules from the training data.

The algorithm repeatedly splits the dataset into smaller subsets based on the feature that provides the best separation between classes.

The final prediction is made at the leaf node after traversing the decision path from the root node.

## 8.8 Hyperparameter Tuning - Decision Tree

Hyperparameter tuning is the process of selecting the best combination of model parameters before training.

A Decision Tree can easily overfit the training data if allowed to grow without restrictions.

By tuning important parameters such as tree depth and minimum samples required for splitting, we can improve the model's ability to generalize to unseen data.

| Hyperparameter    | Purpose                                      |
| ----------------- | -------------------------------------------- |
| criterion         | Split quality measure (`gini` or `entropy`)  |
| max_depth         | Maximum depth of the tree                    |
| min_samples_split | Minimum samples needed to split a node       |

Random Forest is an **ensemble machine learning algorithm** that combines multiple Decision Trees to make a final prediction.

Instead of depending on a single Decision Tree, Random Forest creates many different trees using random samples of the training data and random subsets of features. The predictions from all the trees are then combined using **majority voting** for classification.

In this project, Random Forest is used to predict whether a manufacturing machine will experience a failure.

### Target Variable

* `0` → No Machine Failure
* `1` → Machine Failure

### Why Random Forest?

A single Decision Tree can easily overfit the training data. Random Forest reduces this problem by combining multiple trees, making the model generally more robust and stable.

The main parameter used in our baseline model is:

* `n_estimators=100` → Creates 100 Decision Trees.
* `random_state=42` → Ensures reproducible results.

We will first train a **baseline Random Forest model** and evaluate its performance before performing feature importance analysis and hyperparameter tuning.


## 9.1. Random Forest Baseline Model

````python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Create Random Forest model
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train the model
rf_model.fit(X_train, y_train)

# Predictions
y_train_pred = rf_model.predict(X_train)
y_test_pred = rf_model.predict(X_test)

# Accuracy
train_accuracy = accuracy_score(y_train, y_train_pred)
test_accuracy = accuracy_score(y_test, y_test_pred)

print("Training Accuracy :", train_accuracy)
print("Testing Accuracy  :", test_accuracy)

# Classification Report
print("\nClassification Report:")
print(classification_report(y_test, y_test_pred))

# Confusion Matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_test_pred))
````

Feature importance helps us understand which input variables contributed most to the Random Forest model's predictions.

Random Forest calculates an importance score for each feature based on how much the feature contributes to reducing impurity across the Decision Trees in the forest.

Feature importance is useful not only from a machine learning perspective but also from a manufacturing perspective because it can help identify which machine operating conditions are most associated with predicted failures.

## 9.3.  Hyperparameter Tuning with GridSearchCV

Hyperparameter tuning is used to find a better combination of Random Forest settings.

`GridSearchCV` evaluates different combinations of hyperparameters using cross-validation. In this project, we use the F1-score as the primary scoring metric because the target variable is imbalanced and detecting machine failures is important.

The goal is to find a model that provides a better balance between precision and recall for the machine failure class.

````python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "n_estimators": [100, 200],
    "max_depth": [None, 10, 20],
    "min_samples_split": [2, 5],
    "min_samples_leaf": [1, 2],
    "max_features": ["sqrt", "log2"],
    "class_weight": [None, "balanced"]
}

rf_grid = GridSearchCV(
    estimator=RandomForestClassifier(random_state=42),
    param_grid=param_grid,
    scoring="f1",
    cv=5,
    n_jobs=-1
)

rf_grid.fit(X_train, y_train)
````

## SHAP Explainability

## What is SHAP?

**SHAP (SHapley Additive exPlanations)** is an explainability technique used to understand how machine learning models make predictions.

In our **Predictive Maintenance Project**, SHAP helps us answer:

> **Why did the model predict that a machine is likely to fail?**

SHAP assigns each feature a **SHAP value**, showing how much that feature contributes to the model's prediction.

* **Positive SHAP value** → pushes the prediction toward **Machine Failure (1)**
* **Negative SHAP value** → pushes the prediction toward **Normal Operation (0)**
* **Larger absolute SHAP value** → greater influence on the prediction

### Why SHAP is important in Predictive Maintenance

Accuracy tells us **how well** our model predicts failures, but SHAP helps us understand **why** the model makes those predictions.

This allows us to move from:

**Prediction → Explanation → Root Cause → Maintenance Decision**

In this project, we will use SHAP to explain our **Tuned Decision Tree (`best_dt`)** and identify the machine features that have the greatest influence on failure predictions.

## 13.2 Global SHAP Feature Importance

The first question we want SHAP to answer is:

Which machine features have the greatest overall influence on machine-failure predictions?

We calculate the mean absolute SHAP value for each feature.

Higher value → feature has greater influence on predictions.
We use absolute values because both positive and negative contributions represent influence

````python
# Extract SHAP values for Machine Failure (class 1)
shap_values_failure = shap_values[..., 1]

# Calculate mean absolute SHAP value for each feature
feature_importance = np.abs(shap_values_failure.values).mean(axis=0)

# Create a DataFrame
shap_importance_df = pd.DataFrame({
    "Feature": X_test.columns,
    "Mean_Absolute_SHAP": feature_importance
})

# Sort from most important to least important
shap_importance_df = shap_importance_df.sort_values(
    by="Mean_Absolute_SHAP",
    ascending=False
)

print(shap_importance_df)
````

## 13.2.2 SHAP Beeswarm Plot – Direction and Impact

The **SHAP Beeswarm Plot** helps us understand both the **importance and direction** of each feature's impact on the machine-failure prediction.

Unlike the global SHAP feature-importance plot, which only shows how strongly a feature influences the model, the beeswarm plot shows whether a feature pushes the prediction toward **Machine Failure (1)** or **Normal Operation (0)**.

### How to Interpret the Plot

* **Positive SHAP value (right side)** → pushes the prediction toward machine failure.
* **Negative SHAP value (left side)** → pushes the prediction toward normal operation.
* **Red points** → high feature values.
* **Blue points** → low feature values.
* Greater horizontal spread → stronger influence on the prediction.

### Project Objective

In our Predictive Maintenance project, the beeswarm plot helps us understand the relationship between machine sensor values and failure risk.

For example, it can help determine whether **high torque, high tool wear, or high rotational speed** increases the model's predicted probability of machine failure.

This provides a more detailed understanding of the model and helps connect **ML predictions with manufacturing conditions**.


````python
plt.figure(figsize=(10, 6))

shap.plots.beeswarm(
    shap_values_failure,
    max_display=6,
    show=False
)

plt.title("SHAP Beeswarm Plot - Machine Failure")
plt.tight_layout()
plt.show()
````

* **Air and Process Temperature**

  * Both temperature variables have relatively smaller contributions than torque and tool wear.
  * Their SHAP distributions indicate that operating temperature still contributes to failure predictions, although its influence is comparatively limited in this model.

* **Product Type (`product_type`)**

````python
import pandas as pd

def predict_machine_failure(
    product_type,
    air_temperature_k,
    process_temperature_k,
    rotational_speed_rpm,
    torque_nm,
    tool_wear_min
):
    """
    Predict machine failure using the trained Decision Tree.

    Returns:
        0 = Normal operation predicted
        1 = Machine failure predicted
    """

    # Prepare machine input in the same feature order
    machine_data = pd.DataFrame([{
        "product_type": product_type,
        "air_temperature_k": air_temperature_k,
        "process_temperature_k": process_temperature_k,
        "rotational_speed_rpm": rotational_speed_rpm,
        "torque_nm": torque_nm,
        "tool_wear_min": tool_wear_min
    }])

    # Generate prediction
    prediction = best_dt.predict(machine_data)[0]

    # Display result
    if prediction == 1:
        print("Predicted Status: MACHINE FAILURE")
        print("Recommendation: Inspect the machine and review its operating condition.")
    else:
        print("Predicted Status: NORMAL OPERATION")
        print("Recommendation: Continue routine monitoring and scheduled maintenance.")

    return prediction
````

###  Reusable Machine Failure Prediction Function

A reusable Python function was developed to predict machine failure using six operating parameters: product type, air temperature, process temperature, rotational speed, torque, and tool wear.

The function collects the inputs into a pandas DataFrame, maintains the expected feature order, and passes the data to the trained Decision Tree classifier.

The model returns a binary prediction:

* **0:** Normal operation predicted.
* **1:** Machine failure predicted.

The function demonstrates how a trained machine-learning model can be integrated into a practical predictive-maintenance application.

The prediction should be treated as decision support rather than a guarantee of future machine behavior. Model validation and consistent preprocessing are required before operational deployment.


# <div style="color:white;display:fill;border-radius:5px;background-color:#2C3E50;letter-spacing:0.1px;overflow:hidden"><p style="padding:20px;color:white;overflow:hidden;margin:0;font-size:110%;text-align:center"> 20.Build a Maintenance Decision Framework</p></div>

````python
def maintenance_recommendation(prediction):

    if prediction == 1:
        action = "Priority inspection recommended"
        explanation = (
            "The model predicts a potential machine failure. "
            "Verify the sensor readings and inspect the equipment "
            "according to the plant's maintenance procedure."
        )

    elif prediction == 0:
        action = "Continue routine monitoring"
        explanation = (
            "The model predicts normal operation. "
            "Continue scheduled inspections and routine maintenance."
        )

    else:
        raise ValueError("Prediction must be either 0 or 1.")

    print("Maintenance Action:", action)
    print("Explanation:", explanation)

    return action
````

````python
def assess_machine(
    product_type,
    air_temperature_k,
    process_temperature_k,
    rotational_speed_rpm,
    torque_nm,
    tool_wear_min
):

    # Predict machine condition
    prediction = predict_machine_failure(
        product_type=product_type,
        air_temperature_k=air_temperature_k,
        process_temperature_k=process_temperature_k,
        rotational_speed_rpm=rotational_speed_rpm,
        torque_nm=torque_nm,
        tool_wear_min=tool_wear_min
    )

    # Generate maintenance recommendation
    action = maintenance_recommendation(prediction)

    return {
        "predicted_failure": prediction,
        "maintenance_action": action
    }
````

````python
def predict_machine_failure(
    product_type,
    air_temperature_k,
    process_temperature_k,
    rotational_speed_rpm,
    torque_nm,
    tool_wear_min
):
    # Step 1: Prepare the machine input
    new_machine = pd.DataFrame({
        "product_type": [product_type],
        "air_temperature_k": [air_temperature_k],
        "process_temperature_k": [process_temperature_k],
        "rotational_speed_rpm": [rotational_speed_rpm],
        "torque_nm": [torque_nm],
        "tool_wear_min": [tool_wear_min]
    })

    # Step 2: Predict machine condition
    prediction = loaded_model.predict(new_machine)[0]

    # Step 3: Calculate estimated class probabilities
    probabilities = loaded_model.predict_proba(new_machine)[0]

    no_failure_probability = probabilities[0] * 100
    failure_probability = probabilities[1] * 100

    # Step 4: Display results
    print("=" * 45)
    print("PREDICTIVE MAINTENANCE SYSTEM")
    print("=" * 45)

    print(f"Predicted class: {prediction}")
    print(f"Estimated probability of no failure: {no_failure_probability:.2f}%")
    print(f"Estimated probability of failure:    {failure_probability:.2f}%")

    if prediction == 1:
        print("\nResult: Possible machine failure predicted.")
        print("Action: Investigate the machine and inspect relevant components.")
    else:
        print("\nResult: No machine failure predicted.")
        print("Action: Continue routine monitoring and follow maintenance procedures.")

    print("=" * 45)
````
